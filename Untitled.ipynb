Изучим статистики поля `Quantity`. 

```python
df['Quantity'].describe()
```

Мы видим, что в столбце встречаются отрицательные значения. Попробуем выяснить причину их появления для того, чтобы определиться с дальнейшими действиями.

Для начала проверим, встречаются ли нулевые значения:

```python
df[df['Quantity']==0]
```

Нулевых значений нет. Рассмотрим строки с отрицательным значением `Quantity`:

```python
neg_quantity = df[df['Quantity']<0]
neg_quantity
```

Можно заметить, что среди таких строк есть те, у которых в поле `Price` содержатся нулевые и положительные значения. Проверим, встречаются ли отрицательные:

```python
neg_quantity[neg_quantity['Price']<0]
```

Таких строк не обнаружилось. Попробуем понять, что означает отрицательное значение `Quantity` в сочетании с нулевым содержимым поля `Price`:

```python
neg_quantity[neg_quantity['Price']==0]
```

Мы можем заметить, что содержимое столбца `Description` таких строк не похоже на обычное описание товара. Изучим уникальные записи этого столбца для таких строк:

```python
neg_quantity[neg_quantity['Price']==0]['Description'].unique()
```

По записям можно понять, что в таких строках содержится информация о товарах, которые были возвращены по браку, перепутаны с другими, потеряны, переданы в благотворительные фонды или были иным образом утрачены, не принеся прибыли.

Проверим количество уникальных счетов с такими товарами:

```python
neg_quantity[neg_quantity['Price']==0]['Invoice'].nunique()
```

Количество счетов совпадает с количеством записей с `Quantity`, равным нулю, что в целом согласуется с нашей гипотезой: вероятность получить в одном заказе несколько бракованных товаров невелика.

Эти строки исключать из дальнейшего анализа нельзя: товары с высокой долей списания можно рассматривать как кандидатов на сокращение.

Рассмотрим подробнее записи с ненулевой ценой и отрицательным количеством заказов:

```python
neg_quantity[neg_quantity['Price']>0]
```

Можно заметить, что у такого типа заказов значение `Invoice` начинается с 'C'. Вероятно, что говорит о том, что эти заказы помечены как отменённые (от англ. *cancellation* - отмена). Посмотрим, все ли строки с ненулевой ценой и отрицательным количеством заказов имеют это обозначение в идентификаторе заказа:

```python
neg_quantity[neg_quantity['Price']>0]['Invoice'].str.startswith('C', na=False).sum()
```

Количество строк совпало, то есть, все такие строки имеют 'C' в поле `Invoice`. Также в пользу предположения о том, что ненулевой ценой и отрицательным количеством отмечены строки, принадлежащие в заказам, отменённым полностью, говорит то, что в выведенных строках к одному значению `Invoice` относятся несколько разных товаров. Проверим количество таких заказов и сравним с количеством строк:

```python
neg_quantity[neg_quantity['Price']>0]['Invoice'].nunique()
```

Действительно, на каждый такой заказ приходится в среднем более двух товаров, что косвенно подтверждает гипотезу о том, что строки с отрицательным количеством товаров и положительным значением цены относятся к заказам, отменённым целиком.

Проверим, встречается ли такой тип идентификаторов вне выборки:

```python
mask = (df['Quantity']<0) & (df['Price']>0)
df[~mask]['Invoice'].str.startswith('C', na=False).sum()
```

Вне выборки такое значение встречается один раз. Рассмотрим эту строку:

```python
df[~mask & df['Invoice'].str.startswith('C', na=False)]
```

```python
# 1. bad_items по условию Quantity < 0 & Price == 0
bad_mask = (df['Quantity'] < 0) & (df['Price'] == 0)
bad_items = df[bad_mask]

result = bad_items.groupby('StockCode', as_index=False)['Quantity'].sum()
result['Quantity'] = result['Quantity'] * -1
result = result.rename(columns={'Quantity': 'bad_qty'})

# 2. Остальной датасет (исключая bad_items)
good_sum = (
    df[~bad_mask]
    .groupby('StockCode', as_index=False)['Quantity']
    .sum()
    .rename(columns={'Quantity': 'good_qty'})
)

# 3. Объединение
merged = result.merge(good_sum, on='StockCode', how='left')
merged['good_qty'] = merged['good_qty'].fillna(0)

# 4. Доля  ← вот этот шаг пропущен
merged['ratio'] = np.where(
    merged['good_qty'] > 0,
    merged['bad_qty'] / merged['good_qty'],
    np.nan
)

# 5. Сортировка: товары без продаж (NaN) — наверх
merged['ratio_sort'] = merged['ratio'].fillna(np.inf)

merged = (
    merged
    .sort_values('ratio_sort', ascending=False)
    .drop(columns='ratio_sort')
    .reset_index(drop=True)
)

merged
```
